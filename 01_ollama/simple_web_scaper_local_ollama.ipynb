## What is Ollama

Ollama is a tool that lets you run large language models directly on your own computer. Instead of sending your prompts to a paid online API, the model runs locally on your device. This means you can chat with an AI model while keeping your data on your machine and avoiding per request charges.

### Benefits of running Ollama locally

- No per request API cost  
  You can use the model without paying for tokens or usage fees. The main cost is your own hardware and electricity.

- Works offline  
  After the model is downloaded, you can run it without an internet connection.

- Better privacy and control  
  Prompts and outputs stay on your computer, which can reduce data sharing risks compared with cloud services.

- Easy setup  
  Installing Ollama and pulling a model is usually quick, with simple commands.

- Flexible model choice  
  You can try different models depending on your needs, such as smaller faster models or larger more capable ones.

- Useful for development and testing  
  Local models are convenient for experimenting with prompts, building prototypes, and testing workflows without worrying about usage limits.

- Consistent access  
  Your usage is not affected by service outages, rate limits, or account restrictions on external platforms.

### Simple local workflow

1. Install Ollama on your computer  
2. Download a model using a single command  
3. Run the model and start chatting locally  
4. Optionally connect Ollama to local apps or developer tools for your projects

### USE Ollama for website summarise 


```python
# Create utils function for ollama prompt

# Define our system prompt - you can experiment with this later, changing the last sentence to 'Respond in markdown in Spanish."

system_prompt = """
You are a snarky assistant that analyzes the contents of a website,
and provides a short, snarky, humorous summary, ignoring text that might be navigation related.
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""

# Define our user prompt

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.

"""
```

```python
# See how this function creates exactly the format above

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]


messages_for(url)
```
